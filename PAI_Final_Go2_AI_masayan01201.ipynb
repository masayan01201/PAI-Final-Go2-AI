```python
import genesis as gs
import numpy as np
import matplotlib.pyplot as plt
import imageio.v2 as imageio
from IPython.display import Video, display

# gs.init(backend=gs.gpu)  # ← 今回は実行しない

scene = gs.Scene(
    sim_options=gs.options.SimOptions(
        dt=0.01,
        substeps=2,
    ),
    show_viewer=False,
)

plane = scene.add_entity(
    gs.morphs.Plane()
)

robot = scene.add_entity(
    gs.morphs.URDF(
        file="urdf/go2/urdf/go2.urdf",
        pos=(0.0, 0.0, 0.42),
    )
)

cam = scene.add_camera(
    res=(640, 480),
    pos=(2.6, 2.6, 1.5),
    lookat=(0.0, 0.0, 0.35),
    fov=40,
    GUI=False,
)

scene.build()

print("Scene build OK")
```

```python
joint_names = [
    "FR_hip_joint", "FR_thigh_joint", "FR_calf_joint",
    "FL_hip_joint", "FL_thigh_joint", "FL_calf_joint",
    "RR_hip_joint", "RR_thigh_joint", "RR_calf_joint",
    "RL_hip_joint", "RL_thigh_joint", "RL_calf_joint",
]

dofs_idx = [
    robot.get_joint(name).dofs_idx_local[0]
    for name in joint_names
]

# 安定立位の基準角度
base_pose = np.array([
     0.0, 0.75, -1.45,   # FR
     0.0, 0.75, -1.45,   # FL
     0.0, 0.85, -1.45,   # RR
     0.0, 0.85, -1.45,   # RL
])

robot.set_dofs_kp(
    np.array([45.0] * 12),
    dofs_idx
)

robot.set_dofs_kv(
    np.array([1.5] * 12),
    dofs_idx
)

print("Joint setup OK")
```

```python
frames = []

# まず1秒間、立位を安定させる
for step in range(100):
    robot.control_dofs_position(base_pose, dofs_idx)
    scene.step()

# 約4秒間のトロット動作
for step in range(400):
    t = step * 0.01
    phase = 2.0 * np.pi * 1.5 * t

    # 対角脚ペア
    diagonal_a = np.sin(phase)        # FR + RL
    diagonal_b = np.sin(phase + np.pi) # FL + RR

    target = base_pose.copy()

    # FR
    target[1] += 0.10 * diagonal_a
    target[2] -= 0.15 * max(0.0, diagonal_a)

    # FL
    target[4] += 0.10 * diagonal_b
    target[5] -= 0.15 * max(0.0, diagonal_b)

    # RR
    target[7] += 0.10 * diagonal_b
    target[8] -= 0.15 * max(0.0, diagonal_b)

    # RL
    target[10] += 0.10 * diagonal_a
    target[11] -= 0.15 * max(0.0, diagonal_a)

    robot.control_dofs_position(target, dofs_idx)
    scene.step()

    # ソフトウェアレンダリングが遅いため10ステップごとに記録
    if step % 10 == 0:
        rgb, _, _, _ = cam.render()
        frames.append(rgb)

imageio.mimsave(
    "go2_trot.mp4",
    frames,
    fps=10,
    codec="libx264"
)

print("動画を保存しました:", len(frames), "frames")
display(Video("go2_trot.mp4", embed=True))
```

```python
import genesis as gs
import numpy as np
import matplotlib.pyplot as plt
import imageio.v2 as imageio

from sklearn.ensemble import RandomForestRegressor
from IPython.display import Video, display

scene_ai = gs.Scene(
    sim_options=gs.options.SimOptions(
        dt=0.01,
        substeps=2,
    ),
    show_viewer=False,
)

plane_ai = scene_ai.add_entity(
    gs.morphs.Plane()
)

robot_ai = scene_ai.add_entity(
    gs.morphs.URDF(
        file="urdf/go2/urdf/go2.urdf",
        pos=(0.0, 0.0, 0.42),
        fixed=True,
    )
)

cam_ai = scene_ai.add_camera(
    res=(640, 480),
    pos=(2.5, 2.5, 1.5),
    lookat=(0.0, 0.0, 0.4),
    fov=40,
    GUI=False,
)

scene_ai.build()

print("AI posture scene ready")
```

```python
joint_names_ai = [
    "FR_hip_joint", "FR_thigh_joint", "FR_calf_joint",
    "FL_hip_joint", "FL_thigh_joint", "FL_calf_joint",
    "RR_hip_joint", "RR_thigh_joint", "RR_calf_joint",
    "RL_hip_joint", "RL_thigh_joint", "RL_calf_joint",
]

dofs_idx_ai = [
    robot_ai.get_joint(name).dofs_idx_local[0]
    for name in joint_names_ai
]

# 学習データ
# 入力：希望する姿勢高さ 0.0～1.0
X_train = np.array([
    [0.0],
    [0.25],
    [0.50],
    [0.75],
    [1.0],
])

# 出力：各脚の thigh角度とcalf角度
pose_low = [0.0, 1.15, -2.10]
pose_mid = [0.0, 0.90, -1.75]
pose_high = [0.0, 0.65, -1.35]

def interpolate_pose(height):
    if height <= 0.5:
        ratio = height / 0.5
        leg = np.array(pose_low) * (1-ratio) + np.array(pose_mid) * ratio
    else:
        ratio = (height-0.5) / 0.5
        leg = np.array(pose_mid) * (1-ratio) + np.array(pose_high) * ratio

    return np.tile(leg, 4)

Y_train = np.array([
    interpolate_pose(h[0])
    for h in X_train
])

# AI回帰モデル
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, Y_train)

print("AI model training completed")
print("Model score:", model.score(X_train, Y_train))
```

```python
requested_heights = np.array([
    [0.15],
    [0.40],
    [0.65],
    [0.90],
])

predicted_poses = model.predict(requested_heights)

for height, pose in zip(requested_heights.flatten(), predicted_poses):
    print(f"Requested height = {height:.2f}")
    print(
        "Predicted front-right leg:",
        np.round(pose[:3], 3)
    )
```

```python
import genesis as gs
import numpy as np
import matplotlib.pyplot as plt
import imageio.v2 as imageio

from sklearn.ensemble import RandomForestRegressor
from IPython.display import Video, display
from google.colab import files

# ==========================================
# 1. Genesisシーンの作成
# ==========================================

scene_final = gs.Scene(
    sim_options=gs.options.SimOptions(
        dt=0.01,
        substeps=2,
    ),
    show_viewer=False,
)

# 地面
plane_final = scene_final.add_entity(
    gs.morphs.Plane()
)

# Go2
# 本体は固定するが、後から高さを変更して
# 脚の姿勢と胴体高さを同期させる
robot_final = scene_final.add_entity(
    gs.morphs.URDF(
        file="urdf/go2/urdf/go2.urdf",
        pos=(0.0, 0.0, 0.30),
        fixed=True,
    )
)

# カメラ
cam_final = scene_final.add_camera(
    res=(640, 480),
    pos=(2.3, 2.3, 1.25),
    lookat=(0.0, 0.0, 0.30),
    fov=42,
    GUI=False,
)

scene_final.build()

print("Scene build completed")
```

```python
# ==========================================
# 2. 関節情報
# ==========================================

joint_names_final = [
    "FR_hip_joint", "FR_thigh_joint", "FR_calf_joint",
    "FL_hip_joint", "FL_thigh_joint", "FL_calf_joint",
    "RR_hip_joint", "RR_thigh_joint", "RR_calf_joint",
    "RL_hip_joint", "RL_thigh_joint", "RL_calf_joint",
]

dofs_idx_final = [
    robot_final.get_joint(name).dofs_idx_local[0]
    for name in joint_names_final
]

# ==========================================
# 3. 教師データ
# ==========================================

# AIへの入力
# 0.0：低い姿勢
# 1.0：高い姿勢
X_train = np.array([
    [0.00],
    [0.20],
    [0.40],
    [0.60],
    [0.80],
    [1.00],
])

# 1脚分の関節姿勢
# [hip, thigh, calf]
leg_poses = [
    [0.0, 1.15, -2.10],  # 最も低い
    [0.0, 1.05, -1.95],
    [0.0, 0.95, -1.80],
    [0.0, 0.85, -1.65],
    [0.0, 0.75, -1.50],
    [0.0, 0.65, -1.35],  # 最も高い
]

# 姿勢に合わせた胴体高さ
# 低姿勢では胴体も下げ、高姿勢では上げる
body_heights = [
    0.245,
    0.275,
    0.305,
    0.335,
    0.365,
    0.395,
]

Y_train = []

for leg_pose, body_height in zip(leg_poses, body_heights):

    # 同じ基本姿勢を4脚分に展開
    twelve_joint_angles = np.tile(
        np.array(leg_pose, dtype=float),
        4
    )

    # 12関節＋胴体高さの13出力
    output = np.concatenate([
        twelve_joint_angles,
        [body_height]
    ])

    Y_train.append(output)

Y_train = np.array(Y_train)

print("X_train shape:", X_train.shape)
print("Y_train shape:", Y_train.shape)
```

```python
# ==========================================
# 4. AIモデルの学習
# ==========================================

model_final = RandomForestRegressor(
    n_estimators=200,
    max_depth=6,
    random_state=42
)

model_final.fit(
    X_train,
    Y_train
)

training_score = model_final.score(
    X_train,
    Y_train
)

print("AI model training completed")
print("Training score:", round(training_score, 4))
```

```python
# ==========================================
# 5. 未学習入力に対する推論確認
# ==========================================

test_heights = np.array([
    [0.35],
    [0.55],
    [0.75],
])

test_predictions = model_final.predict(
    test_heights
)

for requested, prediction in zip(
    test_heights.flatten(),
    test_predictions
):
    joint_prediction = prediction[:12]
    body_prediction = prediction[12]

    print("=" * 45)
    print(f"Requested posture level : {requested:.2f}")
    print(
        "Predicted FR joint angles:",
        np.round(joint_prediction[:3], 3)
    )
    print(
        "Predicted body height:",
        round(float(body_prediction), 3),
        "m"
    )
```

```python
# ==========================================
# 6. 関節PD制御
# ==========================================

robot_final.set_dofs_kp(
    np.array([55.0] * 12),
    dofs_idx_final
)

robot_final.set_dofs_kv(
    np.array([2.5] * 12),
    dofs_idx_final
)

print("PD controller setup completed")
```

```python
# ==========================================
# 7. AI姿勢制御と動画記録
# ==========================================

# 姿勢推移
# 低い → 中間 → 高い → 中間 → 低い
motion_levels = [
    0.20,
    0.40,
    0.60,
    0.80,
    0.60,
    0.40,
    0.20,
]

frames_final = []

# 最初の姿勢をAIで推論
first_prediction = model_final.predict(
    np.array([[motion_levels[0]]])
)[0]

current_joint_pose = first_prediction[:12].copy()
current_body_height = float(first_prediction[12])

# 初期姿勢を設定
robot_final.set_dofs_position(
    current_joint_pose,
    dofs_idx_final
)

robot_final.set_pos(
    np.array([
        0.0,
        0.0,
        current_body_height
    ])
)

# 初期姿勢を少し安定させる
for step in range(50):
    robot_final.control_dofs_position(
        current_joint_pose,
        dofs_idx_final
    )
    scene_final.step()

# 各目標姿勢へ移行
for posture_level in motion_levels:

    # AIが目標関節角度と胴体高さを推論
    prediction = model_final.predict(
        np.array([[posture_level]])
    )[0]

    target_joint_pose = prediction[:12]
    target_body_height = float(prediction[12])

    # 60段階で滑らかに遷移
    transition_steps = 60

    start_joint_pose = current_joint_pose.copy()
    start_body_height = current_body_height

    for step in range(transition_steps):

        ratio = (step + 1) / transition_steps

        # イージング処理
        # 始点と終点で動きを緩やかにする
        smooth_ratio = (
            3.0 * ratio**2
            - 2.0 * ratio**3
        )

        interpolated_joint_pose = (
            start_joint_pose * (1.0 - smooth_ratio)
            + target_joint_pose * smooth_ratio
        )

        interpolated_body_height = (
            start_body_height * (1.0 - smooth_ratio)
            + target_body_height * smooth_ratio
        )

        # AIが推論した姿勢へ関節を制御
        robot_final.control_dofs_position(
            interpolated_joint_pose,
            dofs_idx_final
        )

        # 胴体高さも同期して変更
        # 脚が曲がると胴体が下がり、
        # 脚が伸びると胴体が上がる
        robot_final.set_pos(
            np.array([
                0.0,
                0.0,
                interpolated_body_height
            ])
        )

        scene_final.step()

        # 10ステップに1回だけレンダリング
        if step % 10 == 0:
            rgb, _, _, _ = cam_final.render()
            frames_final.append(rgb)

    current_joint_pose = target_joint_pose.copy()
    current_body_height = target_body_height

    # 姿勢を短時間保持
    for hold_step in range(20):

        robot_final.control_dofs_position(
            current_joint_pose,
            dofs_idx_final
        )

        robot_final.set_pos(
            np.array([
                0.0,
                0.0,
                current_body_height
            ])
        )

        scene_final.step()

        if hold_step % 10 == 0:
            rgb, _, _, _ = cam_final.render()
            frames_final.append(rgb)

print("Recorded frames:", len(frames_final))
```

```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------
# AIの入力
# [姿勢レベル, 速度レベル]
# ------------------------------------------

posture_levels_train = [
    0.00,
    0.25,
    0.50,
    0.75,
    1.00,
]

speed_levels_train = [
    0.00,   # ゆっくり
    0.50,   # 普通
    1.00,   # 速い
]

X_train_2d = []
Y_train_2d = []

# ------------------------------------------
# 姿勢レベルから教師データを作る関数
# ------------------------------------------

def make_teacher_output(posture_level, speed_level):

    # 低姿勢から高姿勢までの関節角度
    hip_angle = 0.0

    thigh_angle = (
        1.15
        - 0.50 * posture_level
    )

    calf_angle = (
        -2.10
        + 0.75 * posture_level
    )

    one_leg_pose = np.array([
        hip_angle,
        thigh_angle,
        calf_angle,
    ])

    # 4脚分、12関節へ展開
    twelve_joint_angles = np.tile(
        one_leg_pose,
        4
    )

    # 姿勢レベルに応じた胴体高さ
    body_height = (
        0.235
        + 0.140 * posture_level
    )

    # 速度レベルに応じた遷移ステップ数
    # 遅い：120ステップ
    # 速い：50ステップ
    transition_steps = (
        120
        - 70 * speed_level
    )

    return np.concatenate([
        twelve_joint_angles,
        [
            body_height,
            transition_steps,
        ]
    ])

# ------------------------------------------
# 5姿勢 × 3速度 = 15条件
# ------------------------------------------

for posture_level in posture_levels_train:
    for speed_level in speed_levels_train:

        X_train_2d.append([
            posture_level,
            speed_level,
        ])

        Y_train_2d.append(
            make_teacher_output(
                posture_level,
                speed_level
            )
        )

X_train_2d = np.array(
    X_train_2d,
    dtype=float
)

Y_train_2d = np.array(
    Y_train_2d,
    dtype=float
)

print("教師データ数:", len(X_train_2d))
print("入力形状:", X_train_2d.shape)
print("出力形状:", Y_train_2d.shape)
```

```python
# ------------------------------------------
# Random Forest回帰モデル
# ------------------------------------------

model_2d = RandomForestRegressor(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=1,
    random_state=42
)

model_2d.fit(
    X_train_2d,
    Y_train_2d
)

training_score_2d = model_2d.score(
    X_train_2d,
    Y_train_2d
)

print("AI学習完了")
print(
    "教師データに対するスコア:",
    round(training_score_2d, 4)
)
```

```python
# ------------------------------------------
# 教師データにない条件
# ------------------------------------------

unseen_commands = np.array([
    [0.35, 0.25],
    [0.55, 0.70],
    [0.85, 0.40],
])

unseen_predictions = model_2d.predict(
    unseen_commands
)

for command, prediction in zip(
    unseen_commands,
    unseen_predictions
):
    posture_level = command[0]
    speed_level = command[1]

    joint_angles = prediction[:12]
    body_height = prediction[12]
    transition_steps = prediction[13]

    print("=" * 48)
    print(
        f"未学習入力：姿勢={posture_level:.2f}, "
        f"速度={speed_level:.2f}"
    )
    print(
        "前右脚の推論角度:",
        np.round(joint_angles[:3], 3)
    )
    print(
        "推論した胴体高さ:",
        round(float(body_height), 3),
        "m"
    )
    print(
        "推論した遷移ステップ:",
        int(round(transition_steps))
    )
```

```python
import genesis as gs
import numpy as np
import imageio.v2 as imageio

from PIL import Image, ImageDraw, ImageFont
from IPython.display import Video, display

# ==================================================
# 接地補正付き・提出動画用シーン
# ==================================================

scene_ground = gs.Scene(
    sim_options=gs.options.SimOptions(
        dt=0.01,
        substeps=2,
    ),
    show_viewer=False,
)

plane_ground = scene_ground.add_entity(
    gs.morphs.Plane()
)

robot_ground = scene_ground.add_entity(
    gs.morphs.URDF(
        file="urdf/go2/urdf/go2.urdf",
        pos=(0.0, 0.0, 0.32),

        # 転倒を避けつつ、胴体高さを接地補正で動かす
        fixed=True,
    )
)

cam_ground = scene_ground.add_camera(
    res=(640, 480),
    pos=(2.30, 2.30, 1.20),
    lookat=(0.0, 0.0, 0.28),
    fov=42,
    GUI=False,
)

scene_ground.build()

print("接地補正用シーンを作成しました")
```

```python
# ==================================================
# 12関節の取得
# ==================================================

joint_names_ground = [
    "FR_hip_joint", "FR_thigh_joint", "FR_calf_joint",
    "FL_hip_joint", "FL_thigh_joint", "FL_calf_joint",
    "RR_hip_joint", "RR_thigh_joint", "RR_calf_joint",
    "RL_hip_joint", "RL_thigh_joint", "RL_calf_joint",
]

dofs_idx_ground = [
    robot_ground.get_joint(name).dofs_idx_local[0]
    for name in joint_names_ground
]

# ==================================================
# Go2 URDFではfoot独立リンクがないため
# 各脚の末端であるcalfリンクを使用
# ==================================================

foot_names = [
    "FR_calf",
    "FL_calf",
    "RR_calf",
    "RL_calf",
]

foot_links = []

for name in foot_names:
    link = robot_ground.get_link(name)
    foot_links.append(link)
    print("取得成功:", name)

print("4脚の末端リンクを取得しました")
```

```python
import numpy as np

def get_calf_positions():
    positions = {}

    for name, link in zip(foot_names, foot_links):
        pos = link.get_pos()

        # Tensor → numpy
        if hasattr(pos, "detach"):
            pos = pos.detach().cpu().numpy()

        pos = np.asarray(pos).reshape(-1)

        positions[name] = pos[:3]

    return positions


positions = get_calf_positions()

for name, pos in positions.items():
    print(
        name,
        "X =", round(float(pos[0]), 4),
        "Y =", round(float(pos[1]), 4),
        "Z =", round(float(pos[2]), 4)
    )
```

```python
# ==================================================
# calfリンク高さを利用した接地補正
# ==================================================

REFERENCE_CALF_Z = 0.107


def get_calf_z_values():

    z_values = []

    for link in foot_links:

        pos = link.get_pos()

        if hasattr(pos, "detach"):
            pos = pos.detach().cpu().numpy()

        pos = np.asarray(pos).reshape(-1)

        z_values.append(float(pos[2]))

    return np.array(z_values)


def apply_grounded_pose(
    joint_pose,
    initial_body_height=0.32,
    correction_iterations=4,
):

    # 関節姿勢を設定
    robot_ground.set_dofs_position(
        joint_pose,
        dofs_idx_ground
    )

    body_height = float(initial_body_height)

    for _ in range(correction_iterations):

        robot_ground.set_pos(
            np.array([
                0.0,
                0.0,
                body_height
            ])
        )

        scene_ground.step()

        calf_z = get_calf_z_values()

        # 4脚の平均高さ
        mean_calf_z = float(
            np.mean(calf_z)
        )

        # 基準0.107mとの差
        correction = (
            REFERENCE_CALF_Z
            - mean_calf_z
        )

        body_height += correction

    # 異常値防止
    body_height = float(
        np.clip(
            body_height,
            0.18,
            0.48
        )
    )

    robot_ground.set_pos(
        np.array([
            0.0,
            0.0,
            body_height
        ])
    )

    scene_ground.step()

    final_calf_z = get_calf_z_values()

    return body_height, final_calf_z


print("接地基準:", REFERENCE_CALF_Z, "m")
print("接地補正関数を準備しました")
```

```python
# ==================================================
# calfリンク高さを利用した接地近似補正
# 補正回数：8回
# ==================================================

REFERENCE_CALF_Z = 0.107

def get_calf_z_values():

    z_values = []

    for link in foot_links:

        pos = link.get_pos()

        # Tensor → numpy
        if hasattr(pos, "detach"):
            pos = pos.detach().cpu().numpy()

        pos = np.asarray(pos).reshape(-1)

        z_values.append(float(pos[2]))

    return np.array(z_values)


def apply_grounded_pose(
    joint_pose,
    initial_body_height=0.32,
    correction_iterations=8,   # ← 8回に変更
):

    # AIが推論した関節姿勢を設定
    robot_ground.set_dofs_position(
        joint_pose,
        dofs_idx_ground
    )

    body_height = float(initial_body_height)

    # ==================================================
    # calfリンク高さが基準値0.107mへ近づくまで8回補正
    # ==================================================
    for _ in range(correction_iterations):

        robot_ground.set_pos(
            np.array([
                0.0,
                0.0,
                body_height
            ])
        )

        scene_ground.step()

        calf_z = get_calf_z_values()

        mean_calf_z = float(
            np.mean(calf_z)
        )

        # 基準値との差を計算
        correction = (
            REFERENCE_CALF_Z
            - mean_calf_z
        )

        # 胴体高さを補正
        body_height += correction

    # 異常な高さにならないよう制限
    body_height = float(
        np.clip(
            body_height,
            0.18,
            0.48
        )
    )

    # 最終位置を設定
    robot_ground.set_pos(
        np.array([
            0.0,
            0.0,
            body_height
        ])
    )

    scene_ground.step()

    final_calf_z = get_calf_z_values()

    return body_height, final_calf_z


print("接地基準:", REFERENCE_CALF_Z, "m")
print("補正回数: 8回")
print("接地近似補正関数を準備しました")
```

```python
import numpy as np
import imageio.v2 as imageio

from PIL import Image, ImageDraw, ImageFont
from IPython.display import Video, display


# ==================================================
# 動画設定
# ==================================================

FPS = 4

# 9タスク × 7秒 = 63秒
# タイトル4秒 + 終了5秒
# 合計 約72秒
commands_final = [
    {
        "title": "NORMAL",
        "posture": 0.50,
        "speed": 0.50,
        "unseen": False,
    },
    {
        "title": "LOW / SLOW",
        "posture": 0.00,
        "speed": 0.00,
        "unseen": False,
    },
    {
        "title": "HIGH / FAST",
        "posture": 1.00,
        "speed": 1.00,
        "unseen": False,
    },
    {
        "title": "UNSEEN A",
        "posture": 0.35,
        "speed": 0.25,
        "unseen": True,
    },
    {
        "title": "UNSEEN B",
        "posture": 0.55,
        "speed": 0.70,
        "unseen": True,
    },
    {
        "title": "UNSEEN C",
        "posture": 0.85,
        "speed": 0.40,
        "unseen": True,
    },
    {
        "title": "LOW / FAST",
        "posture": 0.00,
        "speed": 1.00,
        "unseen": False,
    },
    {
        "title": "HIGH / SLOW",
        "posture": 1.00,
        "speed": 0.00,
        "unseen": False,
    },
    {
        "title": "HOME",
        "posture": 0.50,
        "speed": 0.50,
        "unseen": False,
    },
]


# ==================================================
# フレームに説明を追加
# ==================================================

def add_video_text(
    frame,
    task_number,
    title,
    posture,
    speed,
    predicted_height,
    corrected_height,
    calf_z,
    is_unseen,
):
    image = Image.fromarray(
        np.asarray(frame).astype(np.uint8)
    )

    draw = ImageDraw.Draw(image)
    font = ImageFont.load_default()

    # 上部説明パネル
    draw.rectangle(
        [(0, 0), (640, 105)],
        fill=(9, 18, 30)
    )

    condition_text = (
        "UNSEEN COMMAND"
        if is_unseen
        else "TRAINING CONDITION"
    )

    mean_calf_z = float(np.mean(calf_z))
    error_mm = (
        mean_calf_z - REFERENCE_CALF_Z
    ) * 1000.0

    lines = [
        f"Task {task_number}/9 | {title}",
        (
            f"AI input: posture={posture:.2f}, "
            f"speed={speed:.2f}"
        ),
        condition_text,
        (
            f"AI body height={predicted_height:.3f} m | "
            f"Corrected={corrected_height:.3f} m"
        ),
        (
            f"Calf-link Z={mean_calf_z:.3f} m | "
            f"Error={error_mm:+.1f} mm"
        ),
    ]

    y = 8

    for line in lines:
        draw.text(
            (12, y),
            line,
            fill=(255, 255, 255),
            font=font
        )
        y += 19

    return np.asarray(image)


print("動画設定を準備しました")
print("タスク数:", len(commands_final))
```

```python
# ==================================================
# テキスト画面作成
# ==================================================

def make_text_screen(lines, seconds, fps=FPS):

    image = Image.new(
        "RGB",
        (640, 480),
        color=(9, 18, 30)
    )

    draw = ImageDraw.Draw(image)
    font = ImageFont.load_default()

    y = 105

    for line in lines:
        draw.text(
            (65, y),
            line,
            fill=(255, 255, 255),
            font=font
        )
        y += 30

    frame = np.asarray(image)

    return [
        frame.copy()
        for _ in range(int(seconds * fps))
    ]


title_frames = make_text_screen(
    [
        "AI-based Unitree Go2 Posture Control",
        "",
        "AI inputs:",
        "  Posture level + Motion speed",
        "",
        "AI model:",
        "  Random Forest Regressor",
        "",
        "Simulation:",
        "  Genesis 1.4.5",
    ],
    seconds=4,
)


end_frames = make_text_screen(
    [
        "Result",
        "",
        "15 training conditions",
        "3 unseen posture-speed commands",
        "",
        "AI predicted:",
        "  12 joint angles",
        "  Body height",
        "  Transition speed",
        "",
        "8-step calf-link height correction",
        "Simulation completed",
    ],
    seconds=5,
)


print("タイトル画面:", len(title_frames) / FPS, "秒")
print("終了画面:", len(end_frames) / FPS, "秒")
```

```python
# ==================================================
# 8回補正版の最終動画生成
# ==================================================

task_frames = []

# 最初の姿勢を取得
first_command = commands_final[0]

first_prediction = model_2d.predict(
    np.array([[
        first_command["posture"],
        first_command["speed"],
    ]])
)[0]

current_pose = first_prediction[:12].copy()

first_ai_height = float(
    first_prediction[12]
)

current_height, current_calf_z = apply_grounded_pose(
    joint_pose=current_pose,
    initial_body_height=first_ai_height,
    correction_iterations=8,
)

print("初期姿勢を設定しました")


for task_index, command in enumerate(
    commands_final,
    start=1
):
    posture = float(command["posture"])
    speed = float(command["speed"])

    # AIによる推論
    prediction = model_2d.predict(
        np.array([[
            posture,
            speed
        ]])
    )[0]

    target_pose = prediction[:12].copy()

    predicted_body_height = float(
        prediction[12]
    )

    ai_transition_steps = int(
        round(prediction[13])
    )

    # 異常値防止
    transition_steps = int(
        np.clip(
            ai_transition_steps,
            50,
            120
        )
    )

    start_pose = current_pose.copy()

    # 遷移途中の画像を6枚だけレンダリング
    # ソフトウェアレンダリング時間を抑える
    render_points = np.linspace(
        0,
        transition_steps - 1,
        6,
        dtype=int
    )

    transition_images = []

    print(
        f"\nTask {task_index}/9:",
        command["title"]
    )

    print(
        "AI input:",
        f"posture={posture:.2f},",
        f"speed={speed:.2f}"
    )

    print(
        "AI transition steps:",
        transition_steps
    )

    # ==============================================
    # 姿勢間の滑らかな移行
    # ==============================================

    for step in range(transition_steps):

        ratio = (
            step + 1
        ) / transition_steps

        # Smoothstep補間
        smooth_ratio = (
            3.0 * ratio**2
            - 2.0 * ratio**3
        )

        interpolated_pose = (
            start_pose * (1.0 - smooth_ratio)
            + target_pose * smooth_ratio
        )

        # 毎ステップ8回の脚末端高さ補正
        corrected_height, calf_z = apply_grounded_pose(
            joint_pose=interpolated_pose,
            initial_body_height=predicted_body_height,
            correction_iterations=8,
        )

        # 指定した6地点だけレンダリング
        if step in render_points:

            rgb, _, _, _ = cam_ground.render()

            annotated_frame = add_video_text(
                frame=rgb,
                task_number=task_index,
                title=command["title"],
                posture=posture,
                speed=speed,
                predicted_height=predicted_body_height,
                corrected_height=corrected_height,
                calf_z=calf_z,
                is_unseen=command["unseen"],
            )

            transition_images.append(
                annotated_frame
            )

    # ==============================================
    # 遷移部分を3秒にする
    # 3秒 × 4fps = 12フレーム
    # ==============================================

    transition_frame_count = 3 * FPS

    for frame_index in range(
        transition_frame_count
    ):
        source_index = int(
            frame_index
            * len(transition_images)
            / transition_frame_count
        )

        source_index = min(
            source_index,
            len(transition_images) - 1
        )

        task_frames.append(
            transition_images[source_index].copy()
        )

    # ==============================================
    # 最終姿勢を8回補正して4秒保持
    # ==============================================

    final_height, final_calf_z = apply_grounded_pose(
        joint_pose=target_pose,
        initial_body_height=predicted_body_height,
        correction_iterations=8,
    )

    final_rgb, _, _, _ = cam_ground.render()

    final_annotated = add_video_text(
        frame=final_rgb,
        task_number=task_index,
        title=command["title"],
        posture=posture,
        speed=speed,
        predicted_height=predicted_body_height,
        corrected_height=final_height,
        calf_z=final_calf_z,
        is_unseen=command["unseen"],
    )

    # 4秒 × 4fps = 16フレーム
    for _ in range(4 * FPS):
        task_frames.append(
            final_annotated.copy()
        )

    mean_calf_z = float(
        np.mean(final_calf_z)
    )

    error_mm = (
        mean_calf_z
        - REFERENCE_CALF_Z
    ) * 1000.0

    print(
        "Predicted body height:",
        round(predicted_body_height, 4),
        "m"
    )

    print(
        "Corrected body height:",
        round(final_height, 4),
        "m"
    )

    print(
        "Calf-link Z:",
        np.round(final_calf_z, 4)
    )

    print(
        "Reference error:",
        round(error_mm, 2),
        "mm"
    )

    current_pose = target_pose.copy()
    current_height = final_height


print("\n全タスクの生成完了")
print("タスク部分:", len(task_frames) / FPS, "秒")
```

```python
# ==================================================
# タイトル + タスク + 終了画面
# ==================================================

final_video_frames = (
    title_frames
    + task_frames
    + end_frames
)

video_seconds = (
    len(final_video_frames) / FPS
)

print(
    "最終フレーム数:",
    len(final_video_frames)
)

print(
    "最終動画時間:",
    round(video_seconds, 1),
    "秒"
)

if video_seconds < 60:
    print("警告：動画が60秒未満です")
else:
    print("動画時間要件を満たしています")


# ==================================================
# MP4保存
# ==================================================

output_file = (
    "PAI_final_Go2_AI_8step_correction.mp4"
)

imageio.mimsave(
    output_file,
    final_video_frames,
    fps=FPS,
    codec="libx264",
    quality=8,
)

print(
    "動画を保存しました:",
    output_file
)

display(
    Video(
        output_file,
        embed=True,
        width=640
    )
)
```
